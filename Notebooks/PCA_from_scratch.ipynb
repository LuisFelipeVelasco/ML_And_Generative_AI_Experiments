#**PCA FROM SCRATCH**

**PCA is a machine learning algorithm that reduces the dimensions of the data** to avoid problems like :

*   Low learning performance
*   Overfitting
*   Loose of closeness.

The notebook's purpose is implement the PCA algorithm using numpy and native python functions to put in practice the concepts learned it like: matrix operations, projections, eigen values, eigen vectors, variance, covariance,etc.
